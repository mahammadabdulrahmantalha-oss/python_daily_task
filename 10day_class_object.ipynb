```python
#3Create class Book with attributes title, author, price. Add a method display_info()


class Books:
    def __init__(self,title,author,price):

        self.Book_title=title
        self.Book_author=author
        self.Book_price=price

    def display_info(self):
        print(f"Book title  :{self.Book_title}")
        print(f"Book author :{self.Book_author}")
        print(f"Book price  :{self.Book_price}")
        

book=Books("Iron John","Robert Bly",100)
book.display_info()
```

```python
#4Create class Employee with attributes emp_id, name, salary. Create 3 objects. 

class Employee:
    def __init__(self,emp_id, name, salary):

        self.Empolye_ID=emp_id
        self.Employe_name=name
        self.Employe_sly=salary

    def display_info(self):
        print(f"Employe ID     :{self.Empolye_ID}")
        print(f"Employe name   :{self.Employe_name}")
        print(f"Employe slayry : {self.Employe_sly}")
        

emp=Employee("Iron John","Robert Bly",10000)
emp.display_info()
```
