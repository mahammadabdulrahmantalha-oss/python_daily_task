```python
class Person:
    def __init__(self, name, age, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.Person_Name = name
        self.Person_Age = age


class Employee:
    def __init__(self, salary, company, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.salary = salary
        self.Company = company


class Manager(Person, Employee):
    def __init__(self, name, age, salary, company, exp):
        super().__init__(
            name=name,
            age=age,
            salary=salary,
            company=company
        )
        self.Experince = exp

    def Display_info(self):
        print(f"Person Name : {self.Person_Name}")
        print(f"Person Age : {self.Person_Age}")
        print(f"Salary : {self.salary}")
        print(f"Company : {self.Company}")
        print(f"Experience : {self.Experince}")


info = Manager("Talha", 21, 10000, "Google", 10)
info.Display_info()
```
