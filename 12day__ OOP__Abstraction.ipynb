# Abstraction Questions
##### Abstraction Without super() (1–15) 

```python
from abc import ABC , abstractmethod
```

```python
class Shape(ABC):

    #constructor
    
    def __init__(self,area,perimeter):
        self.Area_of_Shape=area
        self.perimeter_of_Shape=perimeter

# absttract:
    @abstractmethod
    def calculate_Area(self):
        pass
    def calculate_perimeter(self):
        pass
# child 1 method
class Circle(Shape):
    #calculater
    def __init__(self,radius):
        self.Radius=radius
#calcu
    def calculate_Area(self):
        print(f"area of the circle      :{3.14*self.Radius**2}")
    def calculate_perimeter(self):
        print(f"perimeter of the circle :{2*3.14*self.Radius}")
#child 2
class Rectangle(ABC):
    
    def __init__(self,lean,wth):

        self.length_of_the_rectangle=lean
        self.width_of_the_rectangle=wth
    def calculate_Area(self):
        print(f"areaof reactangle       : {self.length_of_the_rectangle*self.width_of_the_rectangle}") 
    def calculate_perimeter(self):
        print(f" perimeter of rectangle  :{2*(self.length_of_the_rectangle+self.width_of_the_rectangle)}")

cir1=Circle(5)
cir1.calculate_Area()

cir1=Circle(10)
cir1.calculate_perimeter()
        
        

cir2=Rectangle(5,5)
cir2.calculate_Area()
cir2=Rectangle(5,5)
cir2.calculate_perimeter()
    
```
