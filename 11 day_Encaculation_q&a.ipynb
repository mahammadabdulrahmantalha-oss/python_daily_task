```python
# 1 creat a class named banck acount



# class
 


#class

class Bank_account:
#1 constructor
    
    def __init__(self,bal):
                
        self.__Account_balance=bal

        
#2getter: accessor
    
    def get_balance(self):
        return self.__Account_balance

#3setter:1
    def Deposit(self,amount):
        if amount >0:
           self.__Account_balance=amount+self.__Account_balance  # self.__Account_balance+=amount___ it can also wrrite like this
            

        else:
             
            print("invalid")

#4 setter:2
    def withdraw(self,amount):
        if amount>0 and amount<=self.__Account_balance:

            self.__Account_balance-=amount
            
        else:
            
            print("invalid")
#5  object
bank1=Bank_account(10000)
bank1.get_balance()
        
bank1.Deposit(1000)
bank1.get_balance()

bank1.withdraw(3000)
bank1.get_balance()
```
